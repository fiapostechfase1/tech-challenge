```python
import pandas as pd
```

```python
df = pd.read_csv('ExpVinho.csv', sep='\t')
df
```

```python
colunas = df.columns.tolist()
colunas
```

```python
colunas_anos = [columns for columns in colunas if columns not in ['Id', 'País']]
colunas_anos
```

```python
colunas_qtd = colunas_anos[::2]
colunas_qtd
```

```python
colunas_valor = colunas_anos[1::2]
colunas_valor
```

```python
dados_qtd = df[['País'] + colunas_qtd].melt(id_vars='País', var_name='Ano', value_name='Quantidade')
dados_qtd
```

```python
dados_valor = df[['País'] + colunas_valor].copy()
dados_valor.columns = ['País'] + colunas_qtd
dados_valor = dados_valor.melt(id_vars='País', var_name='Ano', value_name='Valor')
dados_valor.head()
```

```python
df_final = pd.merge(dados_qtd, dados_valor, on=['País', 'Ano'])
df_final
```

```python
df_final['Ano'] = pd.to_numeric(df_final['Ano'], errors='coerce')
max_ano = df_final['Ano'].max()
df_quinze_anos = df_final[df_final['Ano'] >= max_ano - 14]
df_quinze_anos
```

```python
df_quinze_anos.reset_index(drop = True, inplace = True)
df_quinze_anos
```
